```python
import pandas as pd
import altair as alt
import warnings
warnings.filterwarnings("ignore", category=FutureWarning)
```

```python
df = pd.read_csv('../data/data.csv')

df.head()
```

```python
# Take a sample of the data to a size that altair can plot and isn't too busy
sample = df.sample(n=1000, random_state=3) 
```

```python
# Create the selection interaction
selection = alt.selection_interval()

# Plot the first attribute by popularity
chart = alt.Chart(sample).mark_point().add_params(
    selection
).encode(
    x = alt.X("speechiness:Q", title = 'Speechiness'),
    y = alt.Y("popularity:Q", title = 'Popularity'),
    color = alt.condition(selection, alt.value('#EECA3B'), alt.value('gray')),
    opacity = alt.condition(selection, alt.value(1), alt.value(0.3))
)

# Create a title object
text = alt.Chart().mark_text(
    fontSize=15,
    fontWeight=600,
    dx=500,
    dy=0,
).encode(
    text=alt.value("Song Popularity by Musical Attributes")
)

# Plot the title and first chart and then create other charts as a modification of the first chart, changing the X and the color
figure = text&(chart&
        chart.encode(alt.X("danceability:Q", title = 'Danceability'), color = alt.condition(selection, alt.value('#F58517'), alt.value('gray')))|
        chart.encode(alt.X("acousticness:Q", title = 'Acousticness'), color = alt.condition(selection, alt.value('#4C78A8'), alt.value('gray')))&
        chart.encode(alt.X("valence:Q", title = 'Valence'), color = alt.condition(selection, alt.value('#B379A2'), alt.value('gray')))|
        chart.encode(alt.X("liveness:Q", title = 'Liveness'), color = alt.condition(selection, alt.value('#53A24A'), alt.value('gray')))&
        chart.encode(alt.X("instrumentalness:Q", title = 'Instrumentalness'), color = alt.condition(selection, alt.value('#72B7B2'), alt.value('gray'))))

figure
```

```python
# Save figure as html file
figure.save('../docs/Chartfiles/scatter.html')
```
